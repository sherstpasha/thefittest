## Griewank problem with 100 variables

```python
import numpy as np
import plotly.graph_objects as go

from thefittest.benchmarks import Griewank

# Generate grid data for the surface plot
x_plot = np.linspace(-100, 100, 500)
y_plot = np.linspace(-100, 100, 500)
z_plot = Griewank().build_grid(x_plot, y_plot)

# Create a figure and add a surface to it
fig = go.Figure(data=[go.Surface(x=x_plot, y=y_plot, z=z_plot)])

# Show the figure
fig.show()
```

```python
from thefittest.benchmarks import Griewank
from thefittest.optimizers import DifferentialEvolution


n_dimension = 100
left_border = -100.
right_border = 100.
number_of_generations = 500
population_size = 500


optimizer = DifferentialEvolution(fitness_function=Griewank(),
                                  iters=number_of_generations,
                                  pop_size=population_size,
                                  left_border=left_border,
                                  right_border=right_border,
                                  num_variables=n_dimension,
                                  show_progress_each=10,
                                  minimization=True,
                                  mutation="rand_1",
                                  F=0.1,
                                  CR=0.5,
                                  keep_history=True)

optimizer.fit()

fittest = optimizer.get_fittest()
stats = optimizer.get_stats()

print('The fittest individ:', fittest['phenotype'])
print('with fitness', fittest['fitness'])
```

## Custom problem

$$\min_{x_1, x_2} \left( (5 - x_1)^2 + (12 - x_2)^2 \right)$$

```python
from thefittest.optimizers import SHADE


def custom_problem(x):
    return (5 - x[:, 0])**2 + (12 - x[:, 1])**2


n_dimension = 2
left_border = -100.
right_border = 100.
number_of_generations = 100
population_size = 100


optimizer = SHADE(fitness_function=custom_problem,
                  iters=number_of_generations,
                  pop_size=population_size,
                  left_border=left_border,
                  right_border=right_border,
                  num_variables=n_dimension,
                  show_progress_each=10,
                  minimization=True)

optimizer.fit()

fittest = optimizer.get_fittest()

print('The fittest individ:', fittest['phenotype'])
print('with fitness', fittest['fitness'])
```
